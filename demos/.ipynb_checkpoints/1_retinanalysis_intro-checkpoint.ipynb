## Pull Spike Times and Spike Counts

```python
spike_times = ra.get_spike_xarr(response_block, cell_types = cell_types, minimum_n = 3)
spike_counts = xr.apply_ufunc(len, spike_times, vectorize = True)

# Bonus: Use apply_ufunc to pull the average baseline spike count from the 'pre time' portion of the stimulus
def get_avg_baseline(arr, pre_time):
    pre_spikes = [spike for spike in arr if spike < pre_time]

    return len(pre_spikes)

pre_time = epoch_block_params['preTime']
baseline_spike_counts = xr.apply_ufunc(get_avg_baseline, spike_times, 
                                       kwargs = {'pre_time' : pre_time},
                                       vectorize = True)
```

## Plot Some Results

```python
# UPDATED 2026-07-28 (Claude, per yas): replaced the space-constant sweep plot.
# Space constant / stixel size isn't varied within a SpatialNoise block for us (the
# thing that does change epoch-to-epoch is the noise seed, every 2 epochs, which we're
# not summarizing here). Instead, this is a simple per-cell-type spike count summary
# for this single block, as a quick QC check that cells are responding.

# Filter cell types by those that exist in the spike times array
cell_types = sorted(np.unique(spike_times.coords['cell_type'].values))

means = []
sems = []
for ct in cell_types:
    # Pull spike counts and baseline counts for this cell type
    ct_counts = spike_counts.where(spike_counts.cell_type == ct, drop = True)
    ct_baseline = baseline_spike_counts.where(baseline_spike_counts.cell_type == ct, drop = True)

    # Get average baseline across cells and epochs for this cell type
    baseline_mean = ct_baseline.mean(dim = ['cell_id', 'epoch'], skipna = True).item()

    # Average across epochs per cell first (so each cell contributes equally),
    # then take the mean/SEM across cells for the bar + error bar.
    per_cell_mean = ct_counts.mean(dim = 'epoch', skipna = True) - baseline_mean
    n_cells = int(per_cell_mean.count(dim = 'cell_id'))

    means.append(per_cell_mean.mean(dim = 'cell_id', skipna = True).item())
    sems.append(per_cell_mean.std(dim = 'cell_id', skipna = True).item() / np.sqrt(max(n_cells, 1)))

# Plot spike count per cell type
fig, ax = plt.subplots()
x = np.arange(len(cell_types))
ax.bar(x, means, yerr = sems, capsize = 4, color = [f'C{i}' for i in range(len(cell_types))])

ax.axhline(0, color = 'k', linewidth = 0.8)
ax.set_xticks(x)
ax.set_xticklabels(cell_types, rotation = 30, ha = 'right')
ax.set_title(f'Spike Count by Cell Type ({exp_name} {datafile_name})')
ax.set_ylabel('Avg spike count (baseline-subtracted)')
ax.grid(True, axis = 'y')
fig.tight_layout()
```

## Data Quality: EI Footprint Across Light Levels

Checks that the reference chunk's cells are being tracked consistently across NDFs --
not just that `cluster_match()`'s correlation cleared a threshold, but that the actual
EI footprint (electrode-by-electrode amplitude pattern) looks like the same cell at
every light level. For one example cell, shows its EI footprint at the reference chunk
(red, full-array context) plus its EI-matched counterpart at every other NDF this
protocol was run at (zoomed to the same region, matched cell ID + correlation shown in
each panel's title).

Uses the existing, unmodified `cluster_match()`/`ei_corr()` matching convention (same
one used for cell typing and cross-NDF matching everywhere else in this package) --
this is a visualization of that matching, not a stricter or different check. See
`changes/ei_footprint_data_quality_notes.md` for how this compares to yas's MATLAB
`map_ei_amr`/`compare_map_ei_amr` scripts, which use a different matching convention
(stricter correlation cutoff, a minimum-significant-electrode gate, and un-abs'd max
instead of max(abs())) that hasn't been reconciled with `cluster_match()` -- the
correlation values shown here are OUR existing convention, not a MATLAB-matching one.


```python
CELL_ID_TO_CHECK = None  # set to a specific cell_id in analysis_chunk.cell_ids to check that cell instead of auto-picking one

df_ei_matches, ndf_vcds = ra.get_ei_matches_across_ndfs(
    analysis_chunk, exp_name, analysis_chunk.noise_protocol, ss_version=analysis_chunk.ss_version,
)

if len(df_ei_matches) == 0:
    print('No EI matches found against any other NDF for this protocol -- nothing to check.')
else:
    ei_match_summary = ra.summarize_ei_matches(df_ei_matches)
    print(f'{len(ei_match_summary)} ref cell(s) matched at >=1 other NDF (of {len(ndf_vcds)} other NDF(s) found).')
    print('Ranked best-to-worst -- pick a ref_cell_id from the table below, set CELL_ID_TO_CHECK to it, '
          'then just re-run the next cell (no need to re-run this one).')
    display(ei_match_summary.head(20))
```

```python
if len(df_ei_matches) == 0:
    print('No EI matches found against any other NDF for this protocol -- nothing to plot.')
else:
    example_cell_id = (
        CELL_ID_TO_CHECK if CELL_ID_TO_CHECK is not None
        else ra.pick_example_cell_for_footprint(df_ei_matches)
    )
    n_matched = len(df_ei_matches.query('ref_cell_id == @example_cell_id'))
    print(f'Showing EI footprint across NDFs for cell {example_cell_id} '
          f'(matched at {n_matched} of {len(ndf_vcds)} other NDF(s) found).')

    cell_matches = df_ei_matches.query('ref_cell_id == @example_cell_id').set_index('NDF')
    # Highest NDF (dimmest light) first, matches yas's reference figure ordering.
    ndf_values = sorted(ndf_vcds.keys(), reverse=True)

    ndf_labels = [f'NDF {ndf:g}' for ndf in ndf_values]
    panel_vcds = [ndf_vcds[ndf] for ndf in ndf_values]
    panel_cell_ids = [
        int(cell_matches.loc[ndf, 'matched_cell_id']) if ndf in cell_matches.index else None
        for ndf in ndf_values
    ]
    panel_corrs = [
        float(cell_matches.loc[ndf, 'corr']) if ndf in cell_matches.index else None
        for ndf in ndf_values
    ]

    fig = ra.plot_ei_footprint_across_ndfs(
        analysis_chunk.vcd, example_cell_id, f'Reference ({analysis_chunk.chunk_name})',
        ndf_labels, panel_vcds, panel_cell_ids, panel_corrs,
    )
```

### EI footprint mosaic: one cell type, one NDF, scrollable

A different view from the cell above -- instead of one cell across every NDF, this is
every cell of one type at one NDF, so you can scan a whole cell type's EI footprints
for outliers/bad matches in one pass. Reuses `df_ei_matches`/`ndf_vcds` already
computed two cells up (no re-matching, no reloading). Each panel gets its own zoom
window and its own amplitude normalization (it's a different physical cell per panel,
unlike the cell above where every panel is the same cell) -- this checks whether each
footprint looks like a sensible, well-localized signature on its own, not a
cross-cell amplitude comparison.

Displayed in a fixed-height scrollable box (`scrollable_figure()`, plain HTML/CSS --
not Jupyter's `scrolled` cell-output toggle, which was tried elsewhere in this notebook
and reverted for scroll-boxing a whole cell's output rather than just the one figure,
and for not being reliably honored across notebook frontends).


```python
EI_MOSAIC_NDF = sorted(ndf_vcds.keys())[-1] if ndf_vcds else None  # e.g. 5.0 -- set to any NDF value from ndf_vcds (printed below)
EI_MOSAIC_CELL_TYPE = None  # e.g. 'off/brisk transient' -- None picks the first real (non-Unknown/Unmatched) type

if not ndf_vcds:
    print('No NDF data loaded (see the cell above) -- nothing to make a mosaic of.')
else:
    _typing_file_mosaic = next(
        (f for f in analysis_chunk.typing_files if 'classification' in f.lower()),
        analysis_chunk.typing_files[0],
    )
    _col_idx_mosaic = analysis_chunk.typing_files.index(_typing_file_mosaic)
    _available_types_mosaic = sorted(analysis_chunk.df_cell_params[f'typing_file_{_col_idx_mosaic}'].unique())
    _real_types_mosaic = [t for t in _available_types_mosaic if t not in ('Unknown', 'Unmatched')]
    ei_mosaic_cell_type = EI_MOSAIC_CELL_TYPE or (_real_types_mosaic[0] if _real_types_mosaic else _available_types_mosaic[0])
    print(f'EI footprint mosaic: cell type {ei_mosaic_cell_type!r}, NDF {EI_MOSAIC_NDF} '
          f'(available NDFs: {sorted(ndf_vcds.keys())}; available types: {_available_types_mosaic})')

    fig_ei_mosaic = ra.plot_ei_footprint_mosaic_for_cell_type(
        analysis_chunk, df_ei_matches, ndf_vcds, ei_mosaic_cell_type, EI_MOSAIC_NDF,
        typing_file=_typing_file_mosaic,
    )
    ra.scrollable_figure(fig_ei_mosaic, max_height_px=600)
```
