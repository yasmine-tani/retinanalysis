## Find grating datasets

Broad substring search since we don't know the exact protocol name for the DS/OS stimulus (it wasn't recorded anywhere in the pre-extracted file). This will likely also pull in `ContrastResponseGrating` blocks -- that's fine, just pick the right `protocol_name` for the DS/OS one below.

```python
grating_search = ra.get_datasets_from_protocol_names('grating')
print(grating_search['protocol_name'].unique())
display(grating_search)
```

## Choose an experiment + inspect epoch parameters

Defaulting to the example dataset (`20260227A`, `data012`) since that's the one I actually have data for. **Change `ds_os_protocol_name` to whichever exact string showed up above for the drifting-grating DS/OS stimulus** (not `ContrastResponseGrating`).

```python
exp_name = '20260227A'
datafile_name = 'data012'
ds_os_protocol_name = 'manookinlab.protocols.DriftingGrating'  # CHECK/EDIT against grating_search output above

pipeline = ra.create_mea_pipeline(exp_name, datafile_name)
stim_block = pipeline.stim
response_block = pipeline.resp

df_epochs = stim_block.df_epochs
print(f'{len(df_epochs)} epochs found for {exp_name} {datafile_name}')
print('protocol_name(s) present:', df_epochs['protocol_name'].unique())
print()
print('epoch_parameters keys for epoch 0 (VERIFY these match what the code below assumes):')
print(sorted(df_epochs.at[0, 'epoch_parameters'].keys()))
print()
print('epoch 0 orientation/spatialFrequency/temporalFrequency/barWidth:')
for key in ['orientation', 'spatialFrequency', 'temporalFrequency', 'barWidth']:
    print(f'  {key}:', df_epochs.at[0, 'epoch_parameters'].get(key, '<<< NOT FOUND, code below will break >>>'))
```

## Bin spikes and extract F1 per (cell, trial)

`bin_spike_times_at_rate` is protocol-agnostic (unlike the frame-based binning demo 2 uses, which needs stimulus regen) -- it just needs epoch start/end timing, which should exist for any protocol. `b_count=False` gives a rate in Hz, which is what `compute_f1_f0` expects.

100 Hz bins (10ms) give plenty of margin above the ~2-4 Hz temporal frequencies seen in this stimulus -- raise BIN_RATE if your temporalFrequency values are much higher.

```python
BIN_RATE = 100.0  # Hz
response_block.bin_spike_times_at_rate(bin_rate=BIN_RATE, b_count=False)

cell_ids = response_block.cell_ids
n_cells = len(cell_ids)
n_epochs = response_block.n_epochs
print(f'{n_cells} cells x {n_epochs} epochs, binned_spikes shape: {response_block.binned_spikes.shape}')
```

```python
# Build a tidy dataframe: one row per (cell, epoch), with F1/F0 and the stimulus
# condition for that epoch. This is the format the rest of the notebook works from.

rows = []
for j_epoch in range(n_epochs):
    row = df_epochs.iloc[j_epoch]
    if row['protocol_name'] != ds_os_protocol_name:
        continue

    params = row['epoch_parameters']
    pre_bin = int(round(row['preTime'] / 1000.0 * BIN_RATE))
    stim_bins = int(round(row['stimTime'] / 1000.0 * BIN_RATE))
    stim_freq = params['temporalFrequency']

    for i_cell, cell_id in enumerate(cell_ids):
        full_rate = response_block.binned_spikes[i_cell, j_epoch, :]
        stim_rate = full_rate[pre_bin : pre_bin + stim_bins]
        f0, f1 = ra.compute_f1_f0(stim_rate, BIN_RATE, stim_freq)

        rows.append({
            'cell_id': cell_id,
            'epoch_index': j_epoch,
            'orientation': params['orientation'],
            'spatialFrequency': params['spatialFrequency'],
            'temporalFrequency': stim_freq,
            'barWidth': params.get('barWidth'),
            'f0': f0,
            'f1': f1,
        })

df_trials = pd.DataFrame(rows)
print(f'{len(df_trials)} (cell, trial) rows')
display(df_trials.head())
```

## Average F1 across repeats, then compute DSI/OSI per cell

Grouped by (cell, spatialFrequency, temporalFrequency) since this stimulus tests orientation at more than one spatial-frequency/temporal-frequency combination -- a cell can have a different DSI/OSI at each. Averaging is done per orientation across repeats first, then DSI/OSI is computed from that averaged tuning curve.

```python
tuning_curves = (
    df_trials
    .groupby(['cell_id', 'spatialFrequency', 'temporalFrequency', 'orientation'])['f1']
    .mean()
    .reset_index()
)

ds_os_rows = []
for (cell_id, sf, tf), grp in tuning_curves.groupby(['cell_id', 'spatialFrequency', 'temporalFrequency']):
    result = ra.compute_dsi_osi(grp['orientation'].values, grp['f1'].values)
    ds_os_rows.append({
        'cell_id': cell_id,
        'spatialFrequency': sf,
        'temporalFrequency': tf,
        **result,
    })

df_ds_os = pd.DataFrame(ds_os_rows)
display(df_ds_os.sort_values('dsi', ascending=False).head(10))
```

## Plot: example cell's polar tuning curve, and DSI/OSI summary across cells

Picks the highest-DSI cell automatically as the example -- change `example_cell_id`/`example_condition` to look at a specific cell instead.

```python
# --- Example cell polar tuning curve ---
best = df_ds_os.sort_values('dsi', ascending=False).iloc[0]
example_cell_id = best['cell_id']
example_sf = best['spatialFrequency']
example_tf = best['temporalFrequency']

example_curve = tuning_curves.query(
    'cell_id == @example_cell_id and spatialFrequency == @example_sf and temporalFrequency == @example_tf'
).sort_values('orientation')

theta = np.deg2rad(example_curve['orientation'].values)
r = example_curve['f1'].values
# close the loop for plotting
theta_plot = np.append(theta, theta[0])
r_plot = np.append(r, r[0])

fig = plt.figure(figsize=(10, 4.5))
ax1 = fig.add_subplot(121, projection='polar')
ax1.plot(theta_plot, r_plot, marker='o')
ax1.set_title(f'Cell {example_cell_id} (DSI={best["dsi"]:.2f}, OSI={best["osi"]:.2f})\nSF={example_sf:.4f}, TF={example_tf} Hz')

# --- DSI/OSI summary across all cells (for this SF/TF condition) ---
ax2 = fig.add_subplot(122)
summary = df_ds_os.query('spatialFrequency == @example_sf and temporalFrequency == @example_tf')
ax2.scatter(summary['dsi'], summary['osi'], alpha=0.5)
ax2.set_xlabel('DSI')
ax2.set_ylabel('OSI')
ax2.set_xlim(0, 1)
ax2.set_ylim(0, 1)
ax2.set_title(f'All cells, SF={example_sf:.4f}, TF={example_tf} Hz (n={len(summary)})')
ax2.grid(True)

fig.tight_layout()
```
